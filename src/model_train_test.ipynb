```python
from ipynb.fs.full.data_preprocessing import df_train,df_test
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
```

### Preparing list of features and target

```python
X_train=df_train.drop("satisfaction_satisfied",axis=1)
Y_train=df_train["satisfaction_satisfied"]
```

```python
print("Shape of X_train:", X_train.shape)
print("Shape of Y_train:", Y_train.shape)
```

```python
X_test=df_test.drop("satisfaction_satisfied",axis=1)
Y_test=df_test["satisfaction_satisfied"]
```

```python
print("Shape of X_test:", X_test.shape)
print("Shape of Y_test:", Y_test.shape)
```

### Data Scaling

```python
scaler = StandardScaler()

# Fit the scaler on the training data and transform it
X_train_scaled = scaler.fit_transform(X_train)

# Apply the same transformation to the testing data
X_test_scaled = scaler.transform(X_test)
```

```python
print("Feature scaling completed.")
```

```python
print("Scaled training data shape:", X_train_scaled.shape)
print("Scaled testing data shape:", X_test_scaled.shape)
```

### Model Training

```python
model = LogisticRegression(max_iter=100)
model.fit(X_train_scaled, Y_train)
print()
```

```python
print("Logistic Regression model trained successfully!")
```

### Storing Model's Predictions

```python
predict = model.predict(X_test_scaled)
probab=model.predict_proba(X_test_scaled)[:,1]
actual=Y_test
```
