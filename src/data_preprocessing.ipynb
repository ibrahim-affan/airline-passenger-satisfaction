```python
import numpy as np
import pandas as pd
```

```python
df_test=pd.read_csv("test.csv", index_col=0)
df_train=pd.read_csv("train.csv", index_col=0)
```

### Data Preprocessing

Replacing Empty Arrival Delay Values with the Median Value

```python
median_delay = df_train["Arrival Delay in Minutes"].median()
df_train["Arrival Delay in Minutes"] = df_train["Arrival Delay in Minutes"].fillna(median_delay)
df_test["Arrival Delay in Minutes"] = df_test["Arrival Delay in Minutes"].fillna(median_delay)
```

```python
df_train.isnull().sum()
```

```python
df_test.isnull().sum()
```

Removing the id column

```python
df_train=df_train.drop("id",axis=1)
df_test=df_test.drop("id",axis=1)
```

```python
df_train.head()
```

### Encoding

```python
df_train = pd.get_dummies(df_train, drop_first=True)
df_test = pd.get_dummies(df_test, drop_first=True)

print("Dataset shape after encoding:", df_train.shape)
print("Dataset shape after encoding:", df_test.shape)
```
